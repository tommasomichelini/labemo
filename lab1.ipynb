# Attività 1

```python
B = 20
I0 = 400
alpha = 0.90
delta = 5.20
x_min = -5.0
x_max = 5.0

points = np.linspace(x_min, x_max, 161)

def envelope(x):
    return np.sinc(alpha*x/np.pi)**2

def cos2(x):
    return np.cos(delta*x)**2

def mu_function(x):
    return B + I0 * envelope(x) * cos2(x)

print(f"finita: {np.isfinite(mu_function(points)).all()}")
print(f"non negativa: {(mu_function(points) >= 0).all()}")
print(f"simmetrica: {np.allclose(mu_function(points), mu_function(points)[::-1])}")


```
